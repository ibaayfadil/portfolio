```python
# drop id faskes
raw_data = """
12345
12346
12347
"""

ids = [line.strip() for line in raw_data.splitlines() if line.strip()]

with open("katingan.json", "w") as f:
    f.write("[\n")  
    for idx, i in enumerate(ids):
        f.write("  {\n")
        f.write(f'    "id": {i}\n')
        f.write("  }")
        
        if idx < len(ids) - 1:
            f.write(",")
        f.write("\n")   
    f.write("]\n") 
```

```python
import json

raw_data = """
12345
12346
12347
"""

ids = [{"id": int(line.strip())} for line in raw_data.splitlines() if line.strip()]

with open("kati.json", "w") as f:
    json.dump(ids, f, indent=2)
```
